# How to create CLASS & Objects 

##### Typical Syntax for creating Class and objects 

```python
class ClassName:
    # Constructor to initialize attributes
    def __init__(self, attribute1, attribute2):
        self.attribute1 = attribute1
        self.attribute2 = attribute2

    # Method inside class
    def display(self):
        print(f"Attribute 1: {self.attribute1}")
        print(f"Attribute 2: {self.attribute2}")
```

##### Example of class and objects 

```python
class Employees:
    #constructor to initialize the object 
    def __init__(self, salary, age):
        self.salary = salary 
        self.age = age 
    #methods inside the class 
    def diplay(self):
        print(f'employess has salary {self.salary} and age {self.age}')

#creating objects 
emp1 =Employees (20000, 21)
emp2 =Employees(25000, 25)

print(emp1.salary)
print(emp2.salary)

#To see the content of objects 
print(emp1.__dict__)
```

# Constructor and Types of Constructors 

<h3>What is Constructor?</h3>
    <ul>
        <li>A constructor is a special method used to initialize the attributes of a class when an object is created</li>
    <li>In Python, the constructor is defined using the __init__() method</li>
    </ul>
<h3>Types of Constructors </h3>
    <ol> 
        <li>Non-Parameterised Constructors</li>
        <li>Parameterized Constructors</li>
        <li>Default Constructors</li>
    </ol>

##### 1) Non-Parameterized Constructors 

```python
class Employees:
    def __init__(self):
        self.salary = 20000  #default value of Salary 
        self.age = 21        #Default value of Age
emp1 = Employees()
emp2 = Employees()

#To see the Content 
print(emp1.__dict__)
print(emp2.__dict__)
```

##### 2) Parameterized Constructors 

```python
class Employees:
    def __init__(self, salary, age):
        self.salary = salary  
        self.age = age        
emp1 = Employees(20000, 21)
emp2 = Employees(25000, 25)

#To see the Content 
print(emp1.__dict__)
print(emp2.__dict__)
```

##### 3) Defalut Constructors 

```python
class Demo:
    pass  # No constructor defined
obj = Demo()  # Default constructor provided by Python

#------------------Example -------------------------------
class Employees:
    pass 
emp1 = Employees()
print(emp1.__dict__)
```

# Accessing the Class Members 

<h3>What are the class members?</h3>
<p>Class members include:</p>
<ul>
    <li>Attributes (Variables)</li>
    <li>Methods/Actions (Functions defined inside class)</li>
</ul>

<h3>Ways of accessing the class members?</h3>

#### 1) Using Object (Most common way)

```python
class Employees:
    def __init__(self, salary, age):
        self.salary = salary  
        self.age = age       

    def display(self):
        print(f'salary is {self.salary} and age is {self.age}') 
emp1 = Employees(20000, 21)
emp2 = Employees(25000, 25)

#accessing the attributes 
print(emp1.salary) 
print(emp2.age)

#accessing the methods 
print(emp1.display())
```

# Built-in Class FUNCTIONS

<h4>These functions allow you to dynamically access, modify, or delete class attributes </h4>
<ul>
    <li>getattr(object_name, attribute_name default_value)</li>
    <li>setattr(object_name, attribute_name value)</li>
    <li>delattr(object_name, attribute_name)</li>
    <li>hasattr(object_name, attribute_name)</li>
</ul>

```python
class Employees:
    def __init__(self, salary, age):
        self.salary = salary  
        self.age = age       

    def display(self):
        print(f'salary is {self.salary} and age is {self.age}') 
emp1 = Employees(20000, 21)
emp2 = Employees(25000, 25)


#getattr
print(getattr(emp1, 'salary'))

#setattr
print(setattr(emp2, 'salary' , 50000))
print(emp2.__dict__)

#delattr
delattr(emp2, 'age')
print(emp2.__dict__)

#hasattr 
print(hasattr(emp1, 'salary'))
print(hasattr(emp1, 'name'))
```

#### isinstance(object_name, class_name)

```python
class demo:
    pass 
class Employees:
    def __init__(self, salary, age):
        self.salary = salary  
        self.age = age       

    def display(self):
        print(f'salary is {self.salary} and age is {self.age}') 
emp1 = Employees(20000, 21)
emp2 = Employees(25000, 25)

print(isinstance(emp1, Employees))
print(isinstance(emp1, demo))
```

# Built-in Class ATTRIBUTES 

These are system-defined attributes, automatically created when you define a class. They mostly begin and end with double underscores __ (dunder attributes)
<ul>
    <li>__dict__ : Dictionary containing class/object namespace (all members)</li>
    <li>__doc__ : Documentation string for the class (if defined) </li>
    <li>__name__ : Class name (only accessible inside class definition) </li>
    <li>__module__ : Module name where class is defined</li>
    <li>__bases__ : Tuple of base classes (for inheritance)</li>
</ul>

```python
class Employees:
    '''info of salary and age'''
    def __init__(self, salary, age):
        self.salary = salary  
        self.age = age       

    def display(self):
        print(f'salary is {self.salary} and age is {self.age}') 
emp1 = Employees(20000, 21)
emp2 = Employees(25000, 25)


#__dict__ 
print(Employees.__dict__)

#__doc__
print(Employees.__doc__)

#__name__
print(Employees.__name__)

#__mudule__
print(Employees.__module__) #main show it is current module 

#__bases__
print(Employees.__bases__)
```

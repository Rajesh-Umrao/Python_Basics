<h3>➡️What is Nested Class?</h3>
<ul>
    <li>A nested class (also called inner class) is a class defined inside another class.</li>


```python
class University:
    def __init__(self):
        print('Outer class constructor called')
    def display(self):
        print('This is outer display method')
    class College:
        def __init__(self):
            print('Inner class constructor called')
        def show(self):
            print('This is inner show method')
#Creating object 
o1 = University() 
i1 = o1.College() #can'r directly create object for inner cls need to give ref. of outer cls object 

#acessting methods
o1.display()
i1.show()

#REMEMBER: YOU can't access the method of inner class using outer class object 
try:o1.show()
except: print('University object has no attribute show')
```

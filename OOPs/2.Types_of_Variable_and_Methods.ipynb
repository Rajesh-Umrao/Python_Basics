## 1.1) Instance Varibles 

<h4>➡️What is Instance varibale ?</h4>
<ul>
    <li>Instance Variable is a variable that belongs to a specific object (instance) of a class</li>
    <li>Created using self.variable_name inside methods of the class (commonly in __init__) </li>
    <li>Each object has its own copy of instance variables</li>
    <li>Used to store object-specific data</li>
</ul>

```python
class Student:
    def __init__(self, name, marks):
        self.name = name     #instance variable
        self.marks = marks   #instance variable
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)
s3 = Student('Ram', 91)
print(s1.name, s1.marks)
```

### ➡️Object-Specific nature of instance variables 
<p>Since, Each object maintains its own independent copy</p>
<ul>
    <li> Hence Values of variables differs from object to object</li>
    <li> Changing one object’s data doesn’t affect others. For Ex.</li>
        <ul>
            <li>Adding new instance variable for any object</li>
            <li>Changing the value of instance variable </li>
            <li>Deleting the instance var for particular object</li>
        </ul>
</ul>

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)
s3 = Student('Ram', 91)

#1) adding new instance to s1 outside the class 
s1.age = 23
print(s1.__dict__)

# 2) Changing the value of s2 instance 
s2.name = "Veeru"
s2.marks= 100
print(s2.__dict__)

# 3) deleting the marks instance of s2
del s3.marks 
print (s3.__dict__)
```

#### ✅1st Method: Using Constructor  ( __init__ )

```python
class Student:
    def __init__(self, name, marks):  #Constructor 
        self.name = name
        self.marks = marks
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

print(s1.__dict__)  # {'name': 'Akshay', 'marks': 89}
```

#### ✅2nd Method: Using Instance Method

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):      # Instance Method
        print(f"Name: {self.name}, Marks: {self.marks}")

s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

s1.display()  # Name: Rajesh, Marks: 90
```

#### ✅3rd Method: Outside class by using object

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

#creating new instance "age" fro student 1 
s1.age = 24
print(s1.__dict__)
```

## 1.2)Instance Methods 

<h4>➡️What is Instance methods ?</h4>
<ul>
    <li>Functions defined inside a class</li>
    <li>Operate on instance variables</li>
    <li>First parameter is always self, which refers to the current object</li>
    <li>Can access and modify object-specific data</li>
</ul>

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):          #Instance method 
        print(f'{self.name} has {self.marks} marks')
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)
print(s1.display())
```

<h4>➡️Types of Instance methods ?</h4>
<ol>
    <li>Accessor(getter): Used to get the value of instance variables</li>
    <li>Mutator(Setter): Used to modify the value of instance variables</li>
</ol>

##### 1) Accessor(getter)

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def get_marks(self):     
        print(f"{self.name} has{self.marks} marks")
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)
print(s1.get_marks())
```

##### 2)Mutator (setter)

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def update_marks(self, updated_marks):     
        self.marks = updated_marks
s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

s1.update_marks(50)
print(s1.__dict__)
```

<h1 style="color: magenta; font-weight: bold;">2)Class Variables and Class Methods</h1>

## 2.1) Class Varibles 

<h4>➡️What is Class Variables ?</h4>
<ul>
    <li>A Class Variable is a variable that belongs to the class itself, not to any specific object</li>
    <li>It is shared by all objects of the class.</li>
    <li>Defined directly inside the class, outside all methods</li>
    <li>Changing the class variable affects all instances, unless specifically overridden in an object.</li>
</ul>

```python
class Student:
    college ='NIT'  #class variable 
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

#ACCESSING THE CLASS VARIBLE
    #1) BY USING CLASS REFERENCE 
    #2) BY USING OBJECT REFERENCE 

print(Student.college) 
print(s1.college) #accessing using student(object)
```

<h4>➡️Modification of class variables ?</h4>
<ul>
    <li>Can be only done using CLASS REFERENCE </li>
    <li>CAN'T do using Object reference</li>
</ul>

```python
class Student:
    college ='NIT'  #class variable 
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

#Modification using CLASS reference
Student.college = 'NIT Hamirpur'
print(Student.college)

#TRY modification using OBJECT reference 
s1.college = 'NIT_HIMACHAL'
print(s1.college) #it adding object varible 
print(s1.__dict__)

print(s2.college)
print(s2.__dict__)
```

## 2.2) Class Methods 

<h4>➡️What is Class Methods ?</h4>
<ul>
    <li>A Class Method is a method that works on the class itself, not individual objects</li>
    <li>It can access and modify class variables</li>
    <li>Defined using the @classmethod decorator</li>
    <li>First parameter is always cls (refers to the class, similar to how self refers to the object)</li>
</ul>

```python
class Student:
    college ='NIT'  #class variable 
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    @classmethod
    def display(cls):
        print('college Name:', cls.college)

s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

#To access the class: need to give reference of the class 
print(Student.display())         
```

<h4>➡️Modification of Class variables using class methods?</h4>

```python
class Student:
    college ='NIT'  #class variable 
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    @classmethod
    def display(cls):
        cls.college ='NIT Himachal'  #modication of class variable 
        print('college Name:', cls.college)

s1 = Student('Akshay', 89)
s2 = Student('Jay', 94)

#accessing now aftwr
Student.display()
```

<h1 style = "color : magenta; font-weight: bold">3)Static Methods</h1>

<h4>➡️What is Static Methods ?</h4>
<ul>
    <li>Static methods perform operations on external data, meaning data that is not associated with the class or instances directly</li>
    <li>Static methods are defined using the @staticmethod decorator</li>
    <ul>
        <li>They do not require self (instance reference) or cls (class reference) as parameters</li>
        <li>But static methods can access the class variable (if needed )</li>
    </ul>
    <li>Can be called using the class name or object</li>
</ul>

```python
class Bank:
    bank_name ='SBI'
    rate_of_interest =7.5
    @staticmethod
    def simple_interest(principle, years):
        si=(principle*years*Bank.rate_of_interest)/100
        print(si)
print(Bank.simple_interest(10000,5))
```

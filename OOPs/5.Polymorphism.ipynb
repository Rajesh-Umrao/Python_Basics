<h3>➡️What is Polymorphism?</h3>
<ul>
    <li>Polymorphism in Python is the ability of a Python object to take on many forms</li>
    <ul><li>It refers to a situation where a variable, object, function, or method behaves differently depending on the context or the type of object it is interacting with.</li></ul>
<p>Since everything in Python is an object — including variables, functions, data types, and classes — polymorphism can be applied very broadly, making Python highly flexible and dynamic.</p>
</ul>

<h3 style=" color: orange; font-weight: bold">1) Built-in Functional Polymorphism</h3>

<h3>➡️What is Functional Polymorphism?</h3>
<ul>
    <li>Built-in Function Polymorphism refers to the ability of Python's built-in functions (like len(), max(), type(), etc.) to behave differently depending on the type of data they receive. </li>
    <ul><li>One function — many forms of behavior depending on input type.</li></ul>
<li>Why it works?</li><ul><li>Because Python is dynamically typed and uses duck typing, the function adapts its behavior based on the object it receives — without needing inheritance or method overriding.</li></ul>


### __✅len() Function__

```python
print(len("Rajesh"))         # 6 → Number of characters
print(len([1, 2, 3]))         # 3 → Number of elements in list
print(len({"a": 1, "b": 2}))  # 2 → Number of keys in dictionary

# len() works across different types — this is polymorphism
```

### __✅type() Function__

```python
print(type("Hello"))  # <class 'str'>
print(type(123))      # <class 'int'>
print(type([1, 2]))   # <class 'list'>

# ✅ type() adapts to the type of the argument passed.
```

### __✅max() & min() Function__

```python
print(max(1, 5, 9))            # 9
print(max(["a", "c", "b"]))    # "c"
print(max({"a": 1, "b": 3}))   # "b" (based on key names)
```

### __✅sorted() Function__

```python
print(sorted([3, 1, 2]))             # [1, 2, 3]
print(sorted("rajesh"))             # ['a', 'e', 'h', 'j', 'r', 's']
print(sorted({"x": 1, "z": 0}))     # ['x', 'z'] → sorts by keys
```

### __✅Overriding Built-in function's Behavior__

#### Defining a custom Dunder methods: In order to defined own behavior


```python
#This Throw error since it dont have len() function 
class Box:
    def __init__(self, items):
        self.items = items
b = Box(["pen", "pencil", "eraser"])
print(type(b))
try: print(len(b))
except: print('object of type Box has no len()') 
```

```python
#Here we have defined the own version of len() : or dunder method
class Box:
    def __init__(self, items):
        self.items = items
    def __len__(self):
        return len(self.items)
b = Box(["pen", "pencil", "eraser"])
print(type(b))
print(len(b)) #3
```

<h3 style=" color: orange; font-weight: bold">2) Method Overriding (Inheritance Based Polymorphism)</h3>

<h3>➡️What is Method Over-riding?</h3>
<ul>
    <li>Method Overriding occurs when a child (subclass) provides a specific implementation of a method that is already defined in its parent (superclass).</li>
    <ul><li>The method in the child class overrides the method from the parent class — meaning the child’s version is used.
    </li></ul>
</ul>

```python
class Vehicle:
    def __init__(self, name, color, price):
        self.name=name 
        self.color=color
        self.price=price
    def get_details(self):
        print('Name is:', self.name)
        print('Color is:', self.color)
        print('Price is:', self.price)
    def max_speed(self):
        print('Max speed is 100')
    def gears(self):
        print('Nos of gears is 6')

#Suppose you want to make a car class for which init & get details are same but 
# The max speed & gears are different
# So you have to OVERRIDE both max_speed & gears methods 
class Car(Vehicle):
    def max_speed(self):
        print('Max speed is 150')
    def gears(self):
        print('Nos of gears is 7')

#now create the objects 
V1=Vehicle('Truck','Red', 2000000000)
C1=Car('BMW', 'Blue', 7000000)

#Now if you call the getdetails then same method will be executed 
V1.get_details()
C1.get_details()

#But you call the max_speed and gears then 
    #According to object & According to situation the behavior of max_speed is changing 
V1.max_speed()  #100    #max_sppeed method of vehicle is executed 
C1.max_speed()  #150    #max_speed method of CAR is executed 

V1.gears()  #6          #gear method of vehicle is executed      
C1.gears()  #7          #gear methof of car is executed 
```

<h3 style=" color: orange; font-weight: bold">3) Operator Overloading(aka-Operator Polymorphism)</h3>

<h3>➡️What is Operator overloading (aka Operator Polymorphism)?</h3>
<ul>
    <li>Operator polymorphism means the same operator (like +, *, ==) behaves differently depending on the operand's types</li>
<li>Why it works?</li><ul><li>Behind the scenes, Python uses special methods (called dunder methods) to handle operator behavior.</li></ul>


```python
print(5 + 10)         # ➤ 15      (integer addition)
print("Raj" + "esh")  # ➤ "Rajesh" (string concatenation)
print([1] + [2, 3])   # ➤ [1, 2, 3] (list merging)

# ✅ This is polymorphism: 🔁 Same operator (+) → multiple behaviors depending on context
```

#### adding two numbers 

```python
a , b = 10, 5
print(a+b)
print(a.__add__(b))
print(int.__add__(a,b))
```

### 🔴🟡Example: __(+ operator)__ is overloaded for a user defined class

```python
class Book:
    def __init__(self, tittle, pages):
         self.tittle = tittle 
         self.pages = pages 

b1 = Book('Polity', 700)
b2 = Book('History', 500)

try: print('Total nos of pages:', b1+b2) 
except: print('unsupported operand type(s) for +: Book and Book')
    #means in this the + operator has two operand b1 & b2 which has operand data type 'Book'
    #but in the Book class there is no method/function defined for + operator
    #so if you want to add the pages you ahve to defined the add method (see below)
```

```python
class Book:
    def __init__(self, tittle, pages):
         self.tittle = tittle 
         self.pages = pages 
    def __add__(self, others):     #(b1, b2) b1 is in the self and b2 is in others 
        total = self.pages + others.pages  #self.pages go to b1 and others.pages go to b2 object 
        return total                                   
b1 = Book('Polity', 700)
b2 = Book('History', 500)

print('Total nos of pages:', b1+b2) 
```

```python
#if you want that it should return total pages of any nos of book (not only for 2 book)
# Then return the book object instead of returning total pages 
class Book:
    def __init__(self, tittle, pages):
         self.tittle = tittle 
         self.pages = pages 
    def __add__(self, others):     #(b1, b2) b1 is in the self and b2 is in others 
        total = self.pages + others.pages
        return Book('All Books', total)    #retun a book object
    def __str__(self):           
        return str(self.pages)      #whenever you print any object str func is called so we overloaded it                 
b1 = Book('Polity', 700)
b2 = Book('History', 500)
b3 = Book('Geography', 400)

print('Total nos of pages:', b1+b2+b3) 
```

<h3 style=" color: orange; font-weight: bold">4) Method Overloading</h3>

<ul>
    <li>Python does NOT support traditional method overloading like Java or C++.</li>
    <li>In C++ or Java, You can define multiple methods with the same name but different parameters in the same class and the appropriate method will be executed based on the call </li>
    <ul><li>But — Python achieves similar behavior using default arguments, variable-length arguments (*args, **kwargs), and manual checks.</li>
</ul>


<h3>➡️What is Method overloading</h3>
<ul><li>In classical OOP:Method overloading means defining multiple methods with the same name but different numbers or types of parameters or behaviours 
</li></ul>

```python
class Addition:
    def add(self, num1, num2):
        return num1+num2
    def add(self, num1, num2, num3):
        return num1+num2 +num3
a=Addition()
print(a.add(20,30, 20)) #this will work as in py always last defined method is executed 
#BUT IF I TRY TO ADD TWO NOS with thro erro 
try: print(a.add(10,20))
except: print('Addition.add() missing 1 required positional argument: num3')
```

### 🔴Achieving method overloading using different approaches 

```python
#1) USING *args or **kwargs 
class Addition:
    def add(self, *nums):
        return sum(nums)
a=Addition()
print(a.add(20, 20))
print(a.add(20, 20, 10))
print(a.add(20, 20, 10, 50))


#It's not "true" overloading like Java.
#But functionally, you achieve the same goal — 
         #calling the same method name with: Different number of arguments, Different types 
```

<h3 style=" color: orange; font-weight: bold">5) Duck Typing</h3>

<h3>➡️What is Duck Typing</h3>
<ul>
    <li>Duck typing is a concept related to dynamic typing where the type or class of an object is less important than the methods it defines. </li>
    <ul><li>The name comes from the saying: “If it walks like a duck and quacks like a duck, then it probably is a duck."</li>
</ul>


```python
class Duck:
  def swim(self):
    return "Duck swimming"

  def fly(self):
    return "Duck flying"

class Airplane:
  def fly(self):
    return "Airplane flying"

def fly_test(entity):
  print(entity.fly())

# Create objects
duck = Duck()
airplane = Airplane()

# Same function works with different objects
fly_test(duck)      # Output: Duck flying
fly_test(airplane)  # Output: Airplane flying

#In this example, the fly_test() function works with any object that has a fly() method, 
# regardless of its type. This is a typical example of duck typing.
```

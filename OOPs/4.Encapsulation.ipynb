<h3>➡️What is Encapsulation?</h3>
<p>Bundling data (attributes) and methods (functions) that operate on that data into a single unit (class), while restricting direct access to some of the data
</p>
<ul><li>Encapsulation = Grouping + Restriction</li></ul>

### __1)Public__

```python
class Car:
    def __init__(self, speed):
        self.speed=speed 
    def speed(self):
        print(f'Driving at {self.speed} km/h')
#create an object 
c1 =Car(60)
#Accessing variables outside the class via object reference 
c1.speed 
```

### __2)Private__

```python
class Car:
    def __init__(self, speed):
        self.__speed=speed     #Private data 
    def get_speed(self):
        print(f'speed: {self.__speed} km/h')

    def set_speed(self, speed):
        self.__speed = speed
        print(f'speed: {self.__speed} km/h')
#create an object 
c1 =Car(60)

#Accessing outside claas: throw Attributes error
try: c1.speed  
except: print('Attribute Error')

# Accessing speed through get method
c1.get_speed()

#Modifying the speed through Set Methods 
c1.set_speed(1000)
```

### __3)Protected__

```python
class Student:
    def __init__(self):
        self._college = "IIT"  # protected

s = Student()
print(s._college)  # ⚠ Technically accessible, but not recommended
```

<h1 style ="color: magenta; font-weight: bold"> Name Mangling </h1>

```python
class Employee:
    def __init__(self, name, salary):
        self.__name = name      # private
        self.__salary = salary  # private

    def get_salary(self):
        return self.__salary
#Object creation 
e1 = Employee('Yash', 50000)

#Trying to Access Private Variables
    #1) Direct access of data: throw error 
try: e1.__salary 
except: print('Attributes Error')
    #2) Accessing through methods  
print(e1.get_salary())
    #3) Accessing through: _Employee__salary
print(e1._Employee__salary)
print(e1._Employee__name)

#all attributes including name mangling 
print(e1.__dict__) 
```

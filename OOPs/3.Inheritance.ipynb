<h1 style="color: magenta; font-weight : bold"> Inheritance </h1>

<h3 style="font-weight: bold">➡️What is interitance</h3>
<ul>
    <li>Deriving a new class from an existing class so that new class inherits all members(attributes + methods) of existing class is called as inheritance.</li>
</ul>

```python
class Employee:
    company="ABC"
    def display(self):
        print('This is Employee class')
class Manager(Employee):
    bonus=2000
    def show(self):
        print('This is a group of manager')
e1 =Employee()
m1=Manager()

#Accessing:variable and method 
#Remember
    #Child class can access the variable and methods of parent class 
print(m1.company)
print(m1.display())

    # But parent class can not access variables and methods of child class 

try: print(e1.bonus) 
except:print('attribute error') 
try: print(e1.show) 
except:print('attribute error') 
```

<h3 style="font-weight:bold">➡️How constructor works in Inheritance </h3>

<table>
    <tr>
        <th>Situation</th>
        <th>Behavior</th>
    </tr>
    <tr>
        <td>Only Parent has constructor</td>
        <td>If child class has no constructor, Parent constructor runs automatically </td>
    </tr>
    <tr>
        <td>Child has its own constructor</td>
        <td><ul>
            <li>Child constructor overrides parent's constructor</li>
            <li>It means</li>
                    <ul>
                        <li>Child's constructor will run instead of Parent's constructor </li>
                        <li>Parent's constructor won't run automatically unless you explicitly call it using super()</li>
                    </ul>
        </ul></td>
    </tr>
</table>

#### ✅CASE-1: Only Parent has constructor

```python
class father:
    def __init__(self):
        print('father constructor called')
        self.home = '3BHK'
class son(father):
    pass

#create son class object 
s1 = son()
print(s1.__dict__)
```

#### ✅CASE-2: Child has its own constructor

```python
class father:
    def __init__(self):
        print('father constructor called')
        self.home = '3BHK'
class son(father):
    def __init__(self):
        print('Son constructor called')
        self.home = 'VILLA'

#create son class object 
s1 = son()
print(s1.__dict__)
```

#### ✨super() function: with constructor super().__ init __

```python
class Computer:
    def __init__(self, ram, storage):
        self.ram = ram
        self.storage =storage 
        print('computer constructor called')
class Laptop(Computer):
    def __init__(self, ram, storage, model):
        super().__init__(ram, storage)
        self.model =model
        print('laptop constructor called')

l1=Laptop('8gb', '1TB', 'i10')
print(l1.__dict__)
```

#### ✨super() function: with other functions

```python
class Computer:
    def __init__(self, ram, storage):
        self.ram = ram
        self.storage =storage 
        print('computer constructor called')
    def display(self):
        print('welcome to computer world')
class Laptop(Computer):
    def __init__(self, ram, storage, model):
        super().display()
        self.model =model
        print('laptop constructor called')

l1=Laptop('8gb', '1TB', 'i10')
print(l1.__dict__)
```

<h1 style="color: magenta; font-weight : bold">Types of Inheritance </h1>

    <tr>

    </tr>

        <td>Child inherits from Parent, and another class inherits from Child(Grandparent → Parent → Child)</td>

        <td>Hierarchical Inheritance</td>
        <td>Multiple child classes inherit from the same parent class</td>
    </tr>
    <tr>
        <td>Multiple Inheritance</td>
        <td>One child class inherits from multiple parent classes</td>
    </tr>
    <tr>
        <td>Hybrid Inheritance</td>
        <td>Combination of more than one type of inheritance</td>
    </tr>
        <tr>
        <td>Cyclic Inheritance</td>
        <td>Cyclic Inheritance occurs when a class tries to inherit from itself, either directly or indirectly</td>
    </tr>
</table>

### 1️⃣Single Inheritance 

```python
class Father:
    def __init__(self):
         self.sir_name = 'Singh'
class Son(Father):
    def __init__(self, name):
          super().__init__()
          self.name =name 
    def full_name(self):
         print(f'full name: {self.name} {self.sir_name} ')
#Object Creation
f1 =Father()  #parent object
s1 =Son('Yash')  #child object 

#accessing the members
print(s1.__dict__)
print(s1.full_name())

#Remeber:
    #1) Child can call the properties of parent
print(s1.name)
print(s1.sir_name)
print(s1.full_name())
    #2) but vice versa is not true 
print(f1.sir_name)
try: print(f1.full_name())
except: print('attribute error: parent cant access child class ')
```

### 2️⃣Multi-level Inheritance 

```python
class Father:
    def __init__(self):
        self.home ='Banglow'
class Child(Father):
    def __init__(self):
        super().__init__()
        self.flat ='6BHK FLAT'
class Grand_child(Child):
    def __init__(self):
        super().__init__()
        self.car ='BMW'

#creting object 
f1 = Father()
c1 =Child()
gc1 = Grand_child()

#accessing the properties 
    #1) Grand_child can access the element of every calls 
print(gc1.home)
print(gc1.flat)
print(gc1.home)
    #2) Child can acces properties of parent but not Grand child 
print(c1.flat)
print(c1.home)
try: print(c1.car)
except: print('attribute Error')
    #3) Parent can only parent class not Child not grand child
print(f1.home)
try: print(f1.flat)
except: print('attribute Error')
try: print(f1.car)
except: print('attribute Error')
```

### 3️⃣Hierarchical Inheritance

```python
class Computer:
    def __init__(self, ram, storage):
        self.ram=ram 
        self.storage =storage 
    def display(self):
        print('This is computer method')
class HP(Computer):
    def __init__(self, ram, storage, model):
        super().__init__(ram, storage)
        self.model =model
    def show(self):
        print('this is HP laptop method')
class Dell(Computer):
    def __init__(self, ram, storage, processor):
        super().__init__(ram, storage)
        self.processor = processor  

#object creation 
c1 =Computer('8GB', '1TB')
hp1 = HP('8GB', '1TB', 'HP001')
d1 = Dell('8GB', '1TB', 'i10')

#Accessing the properties and methods 
    #1)Child object can access the parent class properties 
print(hp1.model)
print(hp1.ram, hp1.storage)

print(d1.processor)
print(d1.ram, d1.storage)

    #2) Parent class can't access the properties of child class 
try: print(c1.model)
except : print('AttributeError')

    #Child-1 and child-2 cant access the properties of each others
try: print(c1.processor)
except : print('AttributeError')

try: print(d1.model)
except : print('AttributeError')
```

### 4️⃣Multiple Inheritance

```python
class Country:
    def __init__(self):
        print('contry class constructor')
        self.office = 'Delhi'
class State(Country):
    def __init__(self):
        super().__init__()
        print('State class constructor')
        self.office = 'UP'
class District(State, Country):
    def __init__(self):
        super().__init__()
        print('District class constructor')
        self.office = 'Kanpur'

d1 =District()
print(d1.__dict__)
```

### 5️⃣Hybrid Inheritance 

```python
class X:
    pass
class Y:
    pass 
class Z:
    pass
class A(X,Y):
    pass 
class B(Y,Z):
    pass
class M(A,B,Z):
    pass

#finding the mro 
print(X.mro())
print(Y.mro())
print(Z.mro())
print(A.mro())
print(B.mro())
print(M.mro())
```
